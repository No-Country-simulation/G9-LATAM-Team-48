### **6. Cuantificación y Diagnóstico de Datos Faltantes (Valores Nulos)**
Para diseñar la estrategia de limpieza y decidir los métodos de imputación o eliminación de registros, es fundamental mapear el estado de los datos faltantes. Mediante esta línea de código, consolidamos un reporte ordenado ejecutando las siguientes acciones secuenciales:
1. **`.isnull().sum()`**: Identifica las celdas vacías y calcula el total de valores nulos para cada variable.
2. **`.sort_values(ascending=False)`**: Ordena las columnas de mayor a menor cantidad de datos faltantes, colocando al inicio aquellas que requieren atención prioritaria.
3. **`.reset_index(drop=False)`**: Transforma la serie resultante en un DataFrame, rescatando los nombres de las columnas que originalmente actuaban como índices.
4. **`.rename(...)`**: Renombra las columnas con etiquetas descriptivas (`Columnas del dataset` y `Valores nulos`) para generar una tabla final limpia y profesional.


```python
df_EDA_1.isnull().sum().sort_values(ascending=False).reset_index(drop=False).rename(columns={'index':'Columnas del dataset', 0:'Valores nulos'})
```

### **7. Identificación de Registros Duplicados**
La presencia de filas idénticas puede sesgar los análisis estadísticos y alterar el comportamiento de los modelos de aprendizaje automático al sobreponderar observaciones repetidas. 
Para evaluar la integridad del conjunto de datos, combinamos dos métodos de Pandas:
1. **`.duplicated()`**: Analiza fila por fila todo el DataFrame para identificar si existen registros cuyos valores sean exactamente iguales a los de una fila procesada previamente.
2. **`.sum()`**: Contabiliza el número total de verdaderos (`True`) detectados en el paso anterior. 

Este diagnóstico nos permite determinar si es necesario aplicar una depuración de datos mediante la eliminación de redundancias.


```python
df_EDA_1.duplicated().sum()
```

### **8. Análisis Estadístico Descriptivo Global**
Con el propósito de comprender la distribución, dispersión y tendencias centrales de las variables, ejecutamos un resumen estadístico exhaustivo con dos optimizaciones clave:
1. **`include='all'`**: Fuerza a Pandas a analizar tanto las variables numéricas como las categóricas (texto/objetos) en un solo reporte. Para las cuantitativas calcula métricas como la media, desviación estándar, mínimos y máximos; mientras que para las cualitativas identifica el conteo total, valores únicos, la categoría más frecuente (`top`) y su frecuencia (`freq`).
2. **`.T` (Transposición)**: Intercambia las filas por las columnas. Al colocar las variables en el eje vertical y las métricas en el horizontal, se optimiza drásticamente la legibilidad visual en VS Code, especialmente en conjuntos de datos con un alto volumen de dimensiones.


```python
df_EDA_1.describe(include='all').T
```

### **9. Consolidación de Frecuencias para Variables Categóricas**
Con el fin de realizar una auditoría de calidad sobre las variables cualitativas, transformamos y agrupamos los datos para evaluar la consistencia de sus categorías en una sola vista estructurada. El proceso ejecuta las siguientes operaciones en cadena:
1. **`.melt()`**: Transforma el DataFrame de un formato ancho (horizontal) a uno largo (vertical). Concentra los nombres de las columnas seleccionadas en la variable `Variable` y sus correspondientes registros en `Categoría`.
2. **`.value_counts(sort=False).sort_index(...)`**: Contabiliza la frecuencia exacta de aparición de cada par Variable-Categoría, ordenando el índice de forma alfabética inversa para agrupar visualmente las mismas dimensiones.
3. **`.reset_index(...)`**: Aplana la estructura convirtiendo los índices en columnas y asignando el nombre `Cantidad de registros` al conteo final.
4. **`.T` (Transposición)**: Rota la matriz resultante para facilitar la inspección horizontal de todas las frecuencias. Este paso es sumamente crítico aquí, ya que revela problemas de calidad de datos, como inconsistencias de tipografía o duplicación de categorías debido a mayúsculas/minúsculas y variaciones de texto (ej. "urbana costera", "Urbana Costera", "URBANA COSTERA", "sin dato", "s/d").



```python
columnas_cat = [
    'tipo_inmueble',
    'zona',
    'nivel_socioeconomico',
    'tiene_aire_acondicionado',
    'tiene_calentador_agua_electrico',
    'aislamiento_termico',
    'perfil_energetico',
    'tiene_lavadora',
    'fuente_energia_secundaria'
]


resultado_values = (df_EDA_1[columnas_cat].melt(var_name='Variable', value_name='Categoría').value_counts(sort=False).sort_index(ascending=False).reset_index(name='Cantidad de registros'))
pd.set_option('display.max_columns', None)
resultado_values.T
```

### **10. Análisis de Cardinalidad y Valores Únicos (`.nunique()`)**
Para profundizar en la naturaleza de nuestras variables categóricas e identificar identificadores clave, evaluamos la cardinalidad del conjunto de datos. Construimos un reporte estructurado aplicando la siguiente lógica en cadena:
1. **`.nunique()`**: Calcula la cantidad exacta de valores distintos o únicos presentes en cada columna seleccionada.
2. **`.reset_index()`** y **`.rename(...)`**: Convierten la serie resultante en un DataFrame con nombres de columnas claros (`Columna` y `Valores Únicos`) para mejorar la presentación del diagnóstico.
3. **`.sort_values(..., ascending=False)`**: Ordena el reporte de mayor a menor cardinalidad.

Este paso es crítico por dos razones: permite confirmar si la variable `id_registro` funciona correctamente como una llave primaria única (con una cardinalidad alta cercana al total de filas), y expone la presencia de ruidos o inconsistencias de escritura en variables que deberían ser binarias o con pocas categorías (como los valores atípicos de 25 o 16 categorías observados en variables que idealmente son de tipo sí/no o zonas específicas).


```python
columnas_cat = [
    'id_registro',
    'tipo_inmueble',
    'zona',
    'nivel_socioeconomico',
    'tiene_aire_acondicionado',
    'tiene_calentador_agua_electrico',
    'aislamiento_termico',
    'perfil_energetico',
    'tiene_lavadora',
    'fuente_energia_secundaria',
]

df_unicos = (df_EDA_1[columnas_cat].nunique().reset_index().rename(columns={'index': 'Columna', 0: 'Valores Únicos'}).sort_values(by='Valores Únicos', ascending=False).reset_index(drop=True))

print(df_unicos)
```

### **12.Visualización y Análisis de Datos Faltantes (Nulls) graficamente**

Para identificar la presencia y distribución de valores nulos dentro del dataset, se generó un mapa de calor (`sns.heatmap`) mapeando la matriz booleana obtenida con el método `.isnull()`.

Las principales observaciones del gráfico son:
* **Líneas horizontales blancas:** Representan los registros (filas) donde existen datos faltantes (`NaN`) para variables específicas.
* **Distribución de nulos:** Se observa que la pérdida de información no se concentra en un bloque o variable única, sino que está dispersa de forma aleatoria a lo largo de las más de 95,000 observaciones del dataset.
* **Variables afectadas:** Columnas críticas asociadas al consumo energético y hábitos (como `horas_dia_cocina`, `horas_dia_lavanderia`, `antiguedad_electrodomesticos_anios` y `generacion_solar_kwh_mensual`) muestran un patrón punteado constante de datos faltantes.

Este diagnóstico visual confirma la necesidad de aplicar técnicas de imputación (relleno con media, mediana o moda) o evaluar la eliminación de registros incompletos antes de proceder con el cálculo definitivo de la matriz de correlación.


```python
import seaborn as sns
sns.heatmap(df_EDA_1.isnull(), cbar=False)
```

### **13.Análisis de Relaciones y Diagnóstico de Calidad de Datos (Pairplot)**

Se ejecutó un gráfico de pares (`sns.pairplot`) cruzando las variables numéricas con la variable categórica `perfil_energetico` como matiz (`hue`). Este gráfico revela dos hallazgos críticos para la limpieza de la base de datos:

* **Inconsistencia Crítica en las Categorías (Dirty Data):** La leyenda derecha evidencia que la variable `perfil_energetico` tiene múltiples variantes escritas de forma distinta para hacer referencia a un mismo concepto (ej. *'Moderado'*, *'moderado'*, *'MODERADO'* y *'Moderado '* con espacios). 
* **Presencia de Datos Faltantes Ocultos:** Existen múltiples cadenas de texto que representan la ausencia de datos en lugar de nulos reales (ej. *'NO DATA'*, *'sin dato'*, *'s/d'*, *'-'*, *'N/A'*). Esto confunde a los algoritmos y satura la segmentación.
* **Comportamiento de las Variables:** El cruce entre el índice analizado (`Unnamed: 0`) y la variable `cantidad_unidades_aa` (unidades de aire acondicionado) muestra distribuciones concentradas en rangos discretos (valores de 0 a 6 unidades).

Este diagnóstico confirma que, antes de proceder a codificar o correlacionar, es obligatorio **estandarizar el texto** de esta columna mediante técnicas de procesamiento de cadenas (unificar mayúsculas, remover espacios y homologar los valores faltantes).


```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize=(15, 6))
sns.pairplot(df_EDA_1, hue='perfil_energetico')
plt.show()
```
